```go
package main

import (
    "fmt"
    "net/http"
)

func main() {
    resp, err := http.Get("https://google.com")

    if err != nil {
        fmt.Printf("Error: %s\n", err)
        return
    }

    defer resp.Body.Close()
    fmt.Printf("Status: %d\n", resp.StatusCode)
}
```
